```python
reference_date = (
    df_physical_products["InvoiceDate"].max()
    + pd.Timedelta(days=1)
)

print("Reference date:", reference_date)
```

```python
rfm = (
    df_physical_products
    .dropna(subset=["CustomerID"])
    .groupby("CustomerID")
    .agg(
        Recency=("InvoiceDate", lambda x: (
            reference_date - x.max()
        ).days),
        Frequency=("InvoiceNo", "nunique"),
        Monetary=("Revenue", "sum")
    )
)

rfm.head(10).round(2)
```

```python
rfm["R_Score"] = pd.cut(
    rfm["Recency"].rank(pct=True),
    bins=[0, 0.25, 0.50, 0.75, 1],
    labels=[4, 3, 2, 1],
    include_lowest=True
).astype(int)

rfm["F_Score"] = pd.cut(
    rfm["Frequency"].rank(pct=True),
    bins=[0, 0.25, 0.50, 0.75, 1],
    labels=[1, 2, 3, 4],
    include_lowest=True
).astype(int)

rfm["M_Score"] = pd.cut(
    rfm["Monetary"].rank(pct=True),
    bins=[0, 0.25, 0.50, 0.75, 1],
    labels=[1, 2, 3, 4],
    include_lowest=True
).astype(int)

rfm.head(10)
```

```python
rfm["RFM_Total_Score"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

rfm["RFM_Code"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

rfm.sort_values(
    "RFM_Total_Score",
    ascending=False
).head(10)
```

```python
rfm_distribution = (
    rfm["RFM_Code"]
    .value_counts()
    .head(15)
)

print(rfm_distribution)
```

```python
def assign_segment(row):

    r = row["R_Score"]
    f = row["F_Score"]
    m = row["M_Score"]

    if r >= 3 and f >= 3 and m >= 3:
        return "Champions"

    elif f >= 3 and m >= 3:
        return "Loyal Customers"

    elif r >= 3 and f <= 2:
        return "Recent Customers"

    elif r <= 2 and f >= 3:
        return "At Risk"

    elif r <= 2 and f <= 2 and m <= 2:
        return "Lost Customers"

    else:
        return "Other"


rfm["Segment"] = rfm.apply(assign_segment, axis=1)

print(rfm["Segment"].value_counts())
```

```python
segment_analysis = (
    rfm.groupby("Segment")
    .agg(
        Customers=("Monetary", "count"),
        Total_Revenue=("Monetary", "sum"),
        Average_Revenue=("Monetary", "mean"),
        Average_Frequency=("Frequency", "mean"),
        Average_Recency=("Recency", "mean")
    )
    .sort_values("Total_Revenue", ascending=False)
)

segment_analysis["Revenue_Share"] = (
    segment_analysis["Total_Revenue"] /
    segment_analysis["Total_Revenue"].sum()
) * 100

print(segment_analysis.round(2))
```

## Customer Segmentation – RFM Analysis

### 1. Customer distribution

The RFM model identified six customer segments across 4,335 identified customers.

Champions represented 29.97% of the customer base, while Lost Customers accounted for 30.08%.

### 2. Revenue contribution

Champions generated £6,374,332.73, representing 72.77% of total revenue from identified customers.

Loyal Customers contributed 10.33%, followed by Recent Customers with 6.12%.

In contrast, Lost Customers represented 30.08% of the customer base but contributed only 4.24% of revenue.

### 3. Business implications

- Prioritize customer retention strategies for Champions.
- Develop loyalty initiatives to strengthen relationships with Loyal Customers.
- Encourage repeat purchases among Recent Customers.
- Evaluate selective reactivation campaigns for Lost Customers.
- Investigate customer behavior within the Other segment.
